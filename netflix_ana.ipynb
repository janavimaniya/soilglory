```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Separating duration into numeric value and type

data['duration_value'] = data['duration'].str.extract('(\d+)')
data['duration_value'] = data['duration_value'].astype(int)

data['duration_type'] = data['duration'].str.extract('([A-Za-z ]+)')
data['duration_type'] = data['duration_type'].str.strip()
```

```python
data
```

```python
# Age of the content
data['content_age'] = 2024 - data['release_year']

# Number of genres
data['num_genres'] = data['listed_in'].str.split(',').apply(len)

# Title word count
data['title_word_count'] = data['title'].str.split().apply(len)
```
